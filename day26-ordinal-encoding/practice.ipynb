```python
import numpy as np 
import pandas as pd
import seaborn as sns
```

```python
df = pd.read_csv("customer.csv")
df
```

```python
df = df.iloc[: , 2: ]
df
```

```python
from sklearn.model_selection import train_test_split

x_train , x_test , y_train , y_test = train_test_split(df.iloc[: , 0 : 2] , df.iloc[: , -1] , test_size = 0.2 , random_state= 42)
```

```python
x_train.shape , x_test.shape
x_train
```

```python
from sklearn.preprocessing import OrdinalEncoder

encoder = OrdinalEncoder(categories=[
    ["Poor", "Average", "Good"],
    ["School", "UG", "PG"]
])

x_train[["review", "education"]] = encoder.fit_transform(
    x_train[["review", "education"]]
)

x_test[["review", "education"]] = encoder.transform(
    x_test[["review", "education"]]
)
```

```python
oe_x_train = pd.DataFrame(x_train , columns = x_train.columns)
oe_x_test = pd.DataFrame(x_test , columns = x_test.columns)
```

```python
oe_x_train
```

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()

le_y_train = le.fit_transform(y_train)
le_y_test = le.transform(y_test)

le_y_train = pd.DataFrame( le_y_train,  columns=[y_train.name] )

le_y_test = pd.DataFrame( le_y_test, columns=[y_test.name] )
le_y_test
```
